### Importing Libraries and Initializing driver

```python
from selenium import webdriver
from selenium.webdriver.common.by import By
from selenium.webdriver.support.ui import WebDriverWait
from selenium.webdriver.support.ui import Select
from selenium.webdriver.support import expected_conditions as EC

import pandas as pd

driver = webdriver.Edge()
wait = WebDriverWait(driver, 10)
driver.get("https://asc.iitb.ac.in/acadmenu/index.jsp")

input("Login done? \n Press Enter to continue")
```

### Navigating to running courses page

```python
driver.switch_to.default_content()
driver.switch_to.frame("leftPage")

button = driver.find_element(By.ID, "ygtvlabelel1")
button.click()

def forceClick(element):
    driver.execute_script("arguments[0].scrollIntoView(true);", element)
    driver.execute_script("arguments[0].click();", element)

button = wait.until(EC.presence_of_element_located((By.ID, "ygtvlabelel7")))
forceClick(button)

button = wait.until(EC.presence_of_element_located((By.LINK_TEXT, "Running Courses")))
forceClick(button)

driver.switch_to.default_content()
driver.switch_to.frame("rightPage")

```

```python
dropdown = Select(wait.until(EC.presence_of_element_located((By.NAME, "semester"))))
dropdown.select_by_value("1")
wait.until(EC.element_to_be_clickable((By.NAME, "submit"))).click()
```

### Copying Data

```python
Data_1 = []


for branch in range(3, 8):
    branches = wait.until(EC.presence_of_all_elements_located((By.XPATH, "//tr")))
    cols = branches[branch].find_elements(By.TAG_NAME, "td")
    cols[0].find_element(By.TAG_NAME, "a").click()
    rows = wait.until(EC.presence_of_all_elements_located((By.XPATH, "//tr[@bgcolor='#CCCC99']")))

    for row in range(len(rows)):
        rows = wait.until(EC.presence_of_all_elements_located((By.XPATH, "//tr[@bgcolor='#CCCC99']")))
        wait.until(lambda d = driver: len(d.find_elements(By.XPATH, "//tr[@bgcolor='#CCCC99']")) > row)

        iter = rows[row]
        columns = iter.find_elements(By.TAG_NAME, "td")

        typ = columns[1].text.strip()
        code = columns[2].text.strip()
        name = columns[3].text.strip()
        ctype = columns[4].text.strip()
        category = columns[5].text.strip()
        instructor = columns[6].text.strip()
        venue = columns[7].text.strip()
        slot = columns[8].text.strip()
        division = columns[9].text.strip() if len(columns) > 9 else ""

        columns[2].find_element(By.TAG_NAME, "a").click()
        wait.until(EC.presence_of_element_located((By.TAG_NAME, "table")))

        credits = driver.find_element(By.XPATH, "//td[contains(text(),'Total Credits')]/following-sibling::td").text.strip()
        description = driver.find_element(By.XPATH, "//td[contains(text(),'Description')]/following-sibling::td").text.strip()

        driver.back()
        driver.switch_to.default_content()
        driver.switch_to.frame("rightPage")

        Data_1.append({
            "Type": typ,
            "Course Code": code,
            "Course Name": name,
            "Course Type": ctype,
            "Category": category,
            "Instructor": instructor,
            "Venue": venue,
            "Slot": slot,
            "Division": division,
            "Credits": credits,
            "Description": description
        })
    driver.back()
    driver.switch_to.default_content()
    driver.switch_to.frame("rightPage")

sheet_1 = pd.DataFrame(Data_1)
```

### Exporting Data

```python
import re

def clean_excel_text(value):
    
    if isinstance(value, str):
        return re.sub(r'[\x00-\x1F\x7F]', '', value)
    
    return value

sheet_1 = sheet_1.map(clean_excel_text)
sheet_2 = sheet_2.map(clean_excel_text)

with pd.ExcelWriter("25b1242_q_4.xlsx") as writer:

    sheet_1.to_excel(writer, sheet_name="2025_1", index=False)

    sheet_2.to_excel(writer, sheet_name="2025_2", index=False)
```
